```python
import numpy as np
import pandas as pd

df = pd.read_csv("Employee.csv")
print(df.head())

print("\nAvailable columns:", df.columns.tolist())

# Operations on 'Age'
avg_age = np.mean(df["Age"])
print("\nAverage Age:", avg_age)
print("Highest Age:", np.max(df["Age"]))
print("Minimum Age:", np.min(df["Age"]))

# Filtering by 'City' (similar to filtering by 'Department')
print("\nEmployees in Bangalore City:")
print(df[df["City"] == "Bangalore"])

# Add a new column based on existing data, e.g., 'AgeGroup'
df["AgeGroup"] = pd.cut(df["Age"], bins=[20, 30, 40, 50, 60], labels=["20s", "30s", "40s", "50s"])
print("\nDataset with AgeGroup Column:")
print(df.head())

# Sort employees by 'Age' (highest first)
df = df.sort_values("Age", ascending=False)
print("\nEmployees Sorted by Age:")
print(df.head())

# Group by 'Gender' and calculate average 'PaymentTier'
gender_payment_tier = df.groupby("Gender")["PaymentTier"].mean()
print("\nAverage PaymentTier by Gender:")
print(gender_payment_tier)
```
